```python
import torch
import numpy as np
print(f"PyTorch version: {torch.__version__}")

# Check if CUDA (GPU) is available
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
x=torch.zeros(2,2,dtype=torch.float16)
print(x)
print(f"Using shape attribute to print dimensions : {x.shape}")
print(f"Using size method to print dimensions : {x.size()}")
print(f"Using shape, size to access inner dim : {x.shape[1],x.size(1)}")
```

```python
# Directly using list to build a tensor
x=torch.tensor([2.5,0.1])
print(f"{x} and its dtype : {x.dtype}, its shape : {x.shape}")
```

Operations


```python
x=torch.tensor([1,2])
y=torch.tensor([3,4])
print(f"Using Standard addition : {x+y}\nUsing pytorch style add : {torch.add(x,y)}" )
```

```python
x=torch.tensor([1,2])
y=torch.tensor([3,4])
y.add_(x) # inplace addition
print(f"Using inplace addition : {y}")
```

Slicing

```python
x=torch.rand(5,3)
print(f"full tensor: {x}")
print("For slicing we use [row_start:row_stop, col_Start,col_stop]")
print(f"After slicing:\nall rows 1st column: {x[:,0]}\nallcolumns 2nd row: {x[1,:]}\nonly single element: {x[2,2].item()}\nSlicing in between rows and columns:\n{x[2:4,1:3]}")
```

Reshaping


```python
x=torch.rand(4,4)
print(x.shape)
y=x.view(-1,2)
print(y.shape)
```

Numpy conversion

```python
a = torch.ones(5)
print(f"{a} is of dtype - {a.dtype}")
b=a.numpy()
print(f"{b} is of dtype - {b.dtype} which is numpy kind")
```

AUTO Grad


```python
x=torch.randn(3,requires_grad=True)
print(x)
y=x+2
print(y)
z=(y*y).mean()
print(z)

# you want to retain grad for y then
y.retain_grad()
z=z.backward()
print(x.grad)
print(y.grad)
```
