AUTO Grad


```
x=torch.randn(2,3,requires_grad=True)
print(x)
y=x+2
print(y)
z=(y*y).mean()
print(z)

# you want to retain grad for y then
y.retain_grad()
z=z.backward()
print(x.grad)
print(y.grad)
```

What if ys output not a scalar

```
x=torch.tensor([1.0,2.0,3.0], requires_grad=True)
y=torch.stack([x[0]**2,x[0]*x[1],x[1]*x[2]])
v=torch.tensor([1.0,0.1,0.001])
y.backward(v)
x.grad
```

Preventing gradient history

```
x=torch.tensor([1.0,2.0,3.0],requires_grad=True)

#1 - Permanantly false for a tensor

x.requires_grad_(False)
print(x)

# 2 - making a copy of x thats not trackable
x=torch.tensor([1.0,2.0,3.0],requires_grad=True)
y=x.detach()
print(y)

# 3 - a block of code computations that can be skipped tracking
x=torch.tensor([1.0,2.0,3.0],requires_grad=True)
with torch.no_grad():
  y=x+2
print(y)
```

Training example and .zero_grad()

```
weights = torch.ones(4, requires_grad=True)

for epochs in range(3):
  model_output = (weights*3).sum()
  model_output.backward()
  print(weights.grad)
  weights.grad.zero_()
```

#Backprop using example.

```
x=torch.tensor(1.0)
y=torch.tensor(2.0)
w=torch.tensor(1.0,requires_grad=True)

#forward pass

y_hat = w*x
loss = (y_hat-y)**2
print(loss)

# backward pass -- where local grads are calculated automatically

loss.backward()
print(w.grad)
```
